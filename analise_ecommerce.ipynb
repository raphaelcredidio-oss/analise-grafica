```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração estética dos gráficos
sns.set_theme(style="whitegrid")
plt.rcParams.update({'figure.autolayout': True})

# Leitura do arquivo de dados do e-commerce
df = pd.read_csv('ecommerce_estatistica.csv')
df.head()
```

```python
# 1. Gráfico de Histograma - Distribuição de Preços
plt.figure(figsize=(9, 5))
sns.histplot(df['Preço'], kde=True, color='royalblue')
plt.title('Distribuição de Preços dos Produtos', fontsize=14, fontweight='bold')
plt.xlabel('Preço (R$)', fontsize=12)
plt.ylabel('Frequência', fontsize=12)
plt.show()
```

```python
# 2. Gráfico de Dispersão - Relação entre Preço e Nota
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='Preço', y='Nota', hue='Desconto', palette='viridis', alpha=0.8)
plt.title('Relação entre Preço e Nota do Produto', fontsize=14, fontweight='bold')
plt.xlabel('Preço (R$)', fontsize=12)
plt.ylabel('Nota', fontsize=12)
plt.show()
```

```python
# 3. Mapa de Calor - Correlação entre Variáveis Numéricas
plt.figure(figsize=(8, 6))
numeric_cols = ['Nota', 'N_Avaliações', 'Desconto', 'Preço']
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Mapa de Calor - Correlação entre Variáveis Numéricas', fontsize=14, fontweight='bold')
plt.show()
```

```python
# 4. Gráfico de Barra - Top Marcas
plt.figure(figsize=(10, 5))
top_marcas = df['Marca'].value_counts().head(5)
sns.barplot(x=top_marcas.index, y=top_marcas.values, palette='muted')
plt.title('Top 5 Marcas com Maior Número de Produtos', fontsize=14, fontweight='bold')
plt.xlabel('Marca', fontsize=12)
plt.ylabel('Quantidade de Produtos', fontsize=12)
plt.xticks(rotation=15)
plt.show()
```

```python
# 5. Gráfico de Pizza - Distribuição por Gênero
plt.figure(figsize=(7, 7))
genero_counts = df['Gênero'].value_counts()
plt.pie(genero_counts.values, labels=genero_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribuição Percentual por Gênero', fontsize=14, fontweight='bold')
plt.show()
```

```python
# 6. Gráfico de Densidade - Taxa de Desconto
plt.figure(figsize=(9, 5))
sns.kdeplot(df['Desconto'], fill=True, color='darkorange')
plt.title('Densidade da Taxa de Desconto dos Produtos', fontsize=14, fontweight='bold')
plt.xlabel('Desconto (%)', fontsize=12)
plt.ylabel('Densidade', fontsize=12)
plt.show()
```

```python
# 7. Gráfico de Regressão - Preço vs Número de Avaliações
plt.figure(figsize=(9, 5))
sns.regplot(data=df, x='Preço', y='N_Avaliações', scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Regressão: Preço vs Número de Avaliações', fontsize=14, fontweight='bold')
plt.xlabel('Preço (R$)', fontsize=12)
plt.ylabel('Número de Avaliações', fontsize=12)
plt.show()
```
